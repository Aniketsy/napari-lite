# napari-lite in a notebook cell

`Viewer(gui='jupyter')` shows the viewer as widgets in the cell: the canvas (drag to pan, wheel to zoom, napari's keys), a slider per extra dimension, the layer list. The canvas is drawn by the kernel off the screen and streamed to the browser, so this works on a remote server too.

Needs `pip install "napari-lite[jupyter]"`.

```python
import numpy as np

from napari_lite import Viewer

# a bright ball that moves from left to right through 20 slices
z, y, x = np.mgrid[0:20, 0:128, 0:192]
volume = np.exp(-((x - 40 - 6 * z) ** 2 + (y - 64) ** 2) / 300.0)
```

```python
viewer = Viewer(gui='jupyter', title='a notebook viewer')
ball = viewer.add_image(volume, colormap='magma', name='ball')
spots = viewer.add_points(
    [[12, 64, 112], [12, 30, 150]], size=10, face_color='cyan', name='spots'
)
```

The viewer above follows its model, so code changes it too:

```python
viewer.dims.current_step = (12, 0, 0)
viewer.layers['ball'].contrast_limits = (0, 0.8)
```
